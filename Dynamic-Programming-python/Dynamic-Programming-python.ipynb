# Fibonacci Numbers

```python
n = 7

dp = [0] * (n + 1)

dp[0] = 0
dp[1] = 1

for i in range(2, n + 1):
    dp[i] = dp[i - 1] + dp[i - 2]

print(dp[n])
```

## Fibancci series

```python
def feb(n):
    if n == 0:
        return 0

    elif n == 1:
        return 1

    else:
        return feb(n - 1) + feb(n - 2)

print(feb(8))
```

### Assignment 4 — Factorial

```python
def factorial(n):
    if n == 1:
        return 1

    else:
        return n * factorial(n - 1)

factorial(5)
```

### Recursive version

```python
def power(base, exponent):
    if exponent == 0:
        return 1
    else:
        return base * power(base, exponent - 1)

print(power(2, 5))
```

```python
def fib(n, memo={}):
    if n <= 1:
        return n

    if n in memo:
        return memo[n]

    memo[n] = fib(n - 1, memo) + fib(n - 2, memo)

    return memo[n]

print(fib(7))
```

## 1. Basic Knapsack

```python
class Solution:

    def Knapsack(self, W, wt, val, N):

        dp = {}

        def solve(n, cap):

            if n == 0 or cap == 0:
                return 0

            if (n, cap) in dp:
                return dp[(n, cap)]

            cwt = wt[n - 1]
            cv = val[n - 1]

            if cwt <= cap:

                take = cv + solve(n - 1, cap - cwt)

                dont_take = solve(n - 1, cap)

                dp[(n, cap)] = max(take, dont_take)

            else:

                dp[(n, cap)] = solve(n - 1, cap)

            return dp[(n, cap)]


        return solve(N, W)


W = 5
wt = [1, 2, 3]
val = [6, 10, 12]
N = 3

obj = Solution()

result = obj.Knapsack(W, wt, val, N)

print("Maximum value:", result)
```
