# Fibonacci Numbers

```python
n = 7

dp = [0] * (n + 1)

dp[0] = 0
dp[1] = 1

for i in range(2, n + 1):
    dp[i] = dp[i - 1] + dp[i - 2]

print(dp[n])
```

## Fibancci series

```python
def feb(n):
    if n == 0:
        return 0

    elif n == 1:
        return 1

    else:
        return feb(n - 1) + feb(n - 2)

print(feb(8))
```

## Assignment 2 — Count Up

```python
def CountUp(n):
    if n == 10:
        return 

    print(n)
    CountUp(n + 1)

CountUp(0)
```

## Assignment 3 — Sum of Numbers

```python
def Sum_of_numbers(n):
    if n == 0:
        return 0

    elif n == 1:
        return 1

    else:
        return Sum_of_numbers(n + n)

Sum_of_numbers(5)
```
