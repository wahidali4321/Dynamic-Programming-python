# Dynamic Programming python

# Fibonacci Numbers

```python
n = 7

dp = [0] * (n + 1)

dp[0] = 0
dp[1] = 1

for i in range(2, n + 1):
    dp[i] = dp[i - 1] + dp[i - 2]

print(dp[n])
```

## Fibancci series

```python
def feb(n):
    if n == 0:
        return 0

    elif n == 1:
        return 1

    else:
        return feb(n - 1) + (n - 2)

print(feb(8))
```
